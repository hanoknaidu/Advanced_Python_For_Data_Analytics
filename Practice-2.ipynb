## Filtering Using String Methods

```python
df[~df['Product'].str.contains("MacBook")]
```

```python
df[df['Product'].str.contains("MacBook")==False]
```

```python
df[df['Product'].str.contains("MacBook")==True]
```

```python
df[df['Product'].str.contains("MacBook")==False & (df['Memory'].str.contains('SSD'))]
```

```python
df[df['Product'].str.startswith('Mac')  ]
```

```python
df[df['Product'].str.endswith("Book")]
```

```python
df[df['Product'].str.endswith('14')]
```

```python
df[df['Price_usd'].duplicated()]
```

```python
df[df['Product'].str.match('Mac')]
```

```python
df[df['Product'].str.match('Mac.*Air')]
```

## Filtering a DataFrame using SQL

```python
df.query("Price_usd > 1200")
```

```python
df.query("Company == 'Apple' or Company == 'Dell'")
```

```python
df.query("Company == 'Apple' or Price_usd > 1500")
```

```python
import pandasql as psql
```

```python
psql.sqldf("SELECT * FROM df WHERE Company = 'Apple'")
```

```python
psql.sqldf("SELECT Company FROM df WHERE Company = 'Apple' and Inches>14")
```

```python
psql.sqldf("SELECT Company FROM df WHERE Product LIKE '%MacBook%'")
```

```python
psql.sqldf("SELECT * FROM df order by Price_usd DESC")
```

```python
psql.sqldf("SELECT * FROM df order by Price_usd ASC")
```
